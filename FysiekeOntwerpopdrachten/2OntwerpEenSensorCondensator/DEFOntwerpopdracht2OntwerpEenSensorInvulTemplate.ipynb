```python
# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!

import math

# Natuurkundige constante: permittiviteit van het vacuüm
EPSILON_0 = 8.854187e-12 
epsilon_r=3
def vlakkeplaat(A, d, epsilon_r):
    """
    Berekent de capaciteit van een vlakke plaat condensator.
    Input: 
      A: Oppervlakte van de platen in m^2
      d: Afstand tussen de platen (dikte diëlektricum) in meters
      epsilon_r: Relatieve permittiviteit van het materiaal (default = 1 voor lucht)
    Output: 
      C: Capaciteit in Farad (F)
    """
    # De formule: C = (εr * ε0 * A) / d
    C = (epsilon_r * EPSILON_0 * A) / (d*10**-3)  # d wordt omgezet van micrometer naar meter
    return C

# --- TEST JE CODE HIER ---
# Voorbeeld met jouw papiertje:
oppervlakte = 0.01      # 10cm x 10cm
dikte_papier = 0.0001   # 0.1mm
materiaal = 3.0         # εr voor papier

resultaat = vlakkeplaat(oppervlakte, dikte_papier, materiaal)

print(f"De berekende capaciteit is: {resultaat} Farad")
print(f"In picofarad is dat: {resultaat * 1e12:.2f} pF")

# Optioneel: schrijf je functie zodanig dat je arrays als input kan geven en (dus) ook arrays als output.
```

```python
import math

# Constanten (staan vast voor jouw opstelling)
EPSILON_0 = 8.854187e-12
A = 0.01             # 10x10 cm
R = 1000             # 1k Ohm shunt
VIN_MAX = 8          # 8V amplitude
FREQ = 100000        # 100 kHz
OMEGA = 2 * math.pi * FREQ

def bepaal_epsilon(Vout, d_mm):
    """
    Berekent de relatieve permittiviteit (epsilon) op basis van Vout en de huidige dikte.
    Vout: gemeten spanning in Volt
    d_mm: de dikte van de spons op dat moment in mm
    """
    # 1. Bereken d in meters
    d_m = d_mm / 1000
    
    # 2. Bereken de capaciteit C uit de gemeten Vout
    # C = Vout / (R * Vin_max * omega)
    C = Vout / (R * VIN_MAX * OMEGA)
    
    # 3. Bereken epsilon uit C
    # epsilon_r = (C * d) / (epsilon_0 * A)
    epsilon_r = (C * d_m) / (EPSILON_0 * A)
    
    return epsilon_r, C * 1e12

# --- GEBRUIK ---
# Stel je spons is niet ingedrukt (37.5 mm) en je meet een Vout van 0.6 Volt:
gemeten_Vout = 4.56
huidige_dikte = 37.5

eps, cap = bepaal_epsilon(gemeten_Vout, huidige_dikte)

print(f"Bij een meting van {gemeten_Vout}V en dikte {huidige_dikte}mm:")
print(f"-> Berekende Capaciteit: {cap:.2f} pF")
print(f"-> Berekende Epsilon (eps_r): {eps:.2f}")
```

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# Constanten
EPSILON_0 = 8.854187e-12
EPS_SPONS = 15      # Relatieve permittiviteit van de spons
A = 0.01            # Oppervlakte (10x10 cm)

# Bereik van de dikte (in meters)
D_START = 0.04    # 37.5 mm
D_EIND = 0.03     # 27.5 mm (dus 1 cm ingedrukt)

def bereken_C(d):
    """Berekent de capaciteit voor een specifieke afstand d"""
    C = (EPS_SPONS * EPSILON_0 * A) / d
    return C * 1e12  # Omgezet naar pF

# Maak een reeks afstanden aan van start naar eind
diktes = np.linspace(D_START, D_EIND, 100)
capaciteiten = [bereken_C(d) for d in diktes]

# --- PLOTTEN ---
plt.figure(figsize=(8, 5))
plt.plot(diktes * 1000, capaciteiten, color='blue', linewidth=2) # x-as in mm

plt.title("Capaciteit vs. Dikte (Traject: 1 cm indrukking)")
plt.xlabel("Dikte van de spons (d) in mm")
plt.ylabel("Capaciteit (C) in pF")
plt.grid(True, alpha=0.3)

# Omdat we indrukken, is het logisch om de x-as van groot naar klein te zien
plt.gca().invert_xaxis() 

plt.show()

# Print de begin- en eindwaarde voor je verslag
print(f"Capaciteit bij {D_START*1000:.2f} mm: {bereken_C(D_START):.2f} pF")
print(f"Capaciteit bij {D_EIND*1000:.2f} mm: {bereken_C(D_EIND):.2f} pF")
print(min(capaciteiten), max(capaciteiten))
```

```python
# deze is puur voor iteratie maar is naturlijk moeilijk om de precieze details van de spons te weten en te modelleren.
import math
import matplotlib.pyplot as plt
import numpy as np

# Constanten
EPSILON_0 = 8.854187e-12
EPS_SPONS = 15      # Relatieve permittiviteit van de spons
A = 0.01            # Oppervlakte (10x10 cm)
D_START = 0.0375     # Beginhoogte spons (35 mm)
K_SPONS = 1500     # Stijfheid van de spons (N/m) 
G = 9.81            # Zwaartekrachtversnelling

def model_gewicht_naar_capaciteit(grammen):
    # 1. Massa naar Newton (F = m * g)
    massa_kg = grammen / 1000
    kracht_N = massa_kg * G
    
    # 2. Kracht naar indrukking (Wet van Hooke: delta_d = F / k)
    indrukking = kracht_N / K_SPONS
    
    # 3. Nieuwe dikte berekenen (met een 'harde stop' op 1mm dikte)
    d_actueel = max(D_START - indrukking, 0.001)
    
    # 4. Capaciteit berekenen (jouw vlakke plaat formule)
    C = (EPS_SPONS * EPSILON_0 * A) / d_actueel
    return C * 1e12 # Resultaat in pF

# --- TESTPLOT: Gewicht vs Capaciteit ---
gewichten = np.linspace(0, 2000, 100) # Van 0 tot 2kg (2000 gram)
capaciteiten = [model_gewicht_naar_capaciteit(g) for g in gewichten]

plt.figure(figsize=(8, 5))
plt.plot(gewichten, capaciteiten, color='green', linewidth=2)
plt.title("Model: Gewicht op spons vs. Capaciteit")
plt.xlabel("Gewicht (gram)")
plt.ylabel("Capaciteit (pF)")
plt.grid(True, alpha=0.3)

# Voorbeeldpuntje printen
print(f"Bij 500 gram is de voorspelde capaciteit: {model_gewicht_naar_capaciteit(500):.1f} pF")

plt.show()


```

## Opdracht 7: Plot de kalibratiegrafiek.

```python

from scipy.optimize import curve_fit

# Data invoeren
capaciteit_metingen = np.array([34, 55.2, 37.9, 42.8, 66.5, 61.8, 58.2, 72.5, 79])
gewicht_metingen = np.array([0, 393, 22.2, 111, 750, 596, 465, 1000, 1500])

# Sorteren van de data op gewicht (belangrijk voor een strakke lijn)
indices = np.argsort(gewicht_metingen)
x_data = gewicht_metingen[indices]
y_data = capaciteit_metingen[indices]

# De fit functie
def log_functie(x, a, b, c):
    return a * np.log(x + b) + c

# we schatten startwaarden: 
# b moet groter zijn dan 0 om log(0) te voorkomen bij x=0
p0 = [10, 10, 30] 

# Uitvoeren van de fit
popt, pcov = curve_fit(log_functie, x_data, y_data, p0=p0)


# Plotten
fit_gewicht = np.linspace(0, 1600, 100)
plt.figure(figsize=(10, 6))

plt.plot(x_data, y_data, 'o', label='Gemeten data (pF)')
plt.plot(fit_gewicht, log_functie(fit_gewicht, *popt), 'r--',
         label=f'Fit: {popt[0]:.2f} * ln(x + {popt[1]:.2f}) + {popt[2]:.2f}')

plt.xlabel('Gewicht ($g$)')
plt.ylabel('Capaciteit ($pF$)')
plt.title('Kalibratiegrafiek: Gewicht vs. Capaciteit')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
